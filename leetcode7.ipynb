```python
class Solution(object):
    def lengthOfLongestSubstring(self, s):
        """
        :type s: str
        :rtype: int
        """
        lastSeen = {}
        left = 0
        maxLength = 0

        for right in range(len(s)):
            char = s[right]

            if char in lastSeen and lastSeen[char] >= left:
                left = lastSeen[char] + 1

            lastSeen[char] = right

            maxLength = max(maxLength, right - left + 1)

        return maxLength
        
```
